### Cleaning and constructing Key Measurables

Injuries and robustness to destruction are a key interest point for the client. Clean and impute relevant columns and then create derived fields that best quantifies what the client wishes to track. **Use commenting or markdown to explain any cleaning assumptions as well as any derived columns you create.**

**Construct metric for fatal/serious injuries**

*Hint:* Estimate the total number of passengers on each flight. The likelihood of serious / fatal injury can be estimated as a fraction from this.

```python
# Injury-related columns
injury_cols = [
    'Total.Fatal.Injuries',
    'Total.Serious.Injuries',
    'Total.Minor.Injuries',
    'Total.Uninjured'
]

# Fill NaN values with 0 across these columns
df[injury_cols] = df[injury_cols].fillna(0)
```

```python
print(df[injury_cols].isna().sum())
```

```python
# Sum across the 4 columns for each row to get total occupants on board
df['Total_on_board'] = df[injury_cols].sum(axis=1)
```

```python

# Compute severe injury fraction guarded against division by zero
df['Severe_Injury_Fraction'] = np.where(
    df['Total_on_board'] > 0,
    (df['Total.Fatal.Injuries'] + df['Total.Serious.Injuries']) / df['Total_on_board'],
    np.nan
)
```

```python
# Inspect summary statistics and check that values range between 0 and 1
print(df[['Total_on_board', 'Severe_Injury_Fraction']].describe())
```

### Investigate the *Make* column
- Identify cleaning tasks here
- List cleaning tasks clearly in markdown
- Execute the cleaning tasks
- For your analysis, keep Makes with a reasonable number (you can put the threshold at 50 though lower could work as well)

```python
df['Make'] = df['Make'].astype(str).str.strip().str.upper()
```

```python
# Define dictionary mapping synonyms and variants to standard names
make_mapping = {
    'MCDONNELL-DOUGLAS': 'MCDONNELL DOUGLAS',
    'MCDONNELL DOUGLAS AIRCRAFT CO': 'MCDONNELL DOUGLAS',
    'MCDONNELL DOUGLAS CORPORATION': 'MCDONNELL DOUGLAS',
    'THE BOEING COMPANY': 'BOEING',
    'BOEING COMPANY': 'BOEING',
    'AIRBUS INDUSTRIE': 'AIRBUS',
    'AIRBUS INDUSTRIES': 'AIRBUS',
    'DEHAVILLAND': 'DE HAVILLAND',
    'BEECHCRAFT': 'BEECH',
    'BEECH AIRCRAFT': 'BEECH',
    'BEECH AIRCRAFT CORPORATION': 'BEECH',
    'CESSNA AIRCRAFT CO': 'CESSNA',
    'CESSNA AIRCRAFT COMPANY': 'CESSNA',
    'PIPER AIRCRAFT': 'PIPER',
    'PIPER AIRCRAFT INC': 'PIPER',
    'PIPER AIRCRAFT CORPORATION': 'PIPER'
}

# Apply the mapping
df['Make'] = df['Make'].replace(make_mapping)
```

```python
# Count occurrences of each cleaned make
counts = df['Make'].value_counts()
```

```python
# Get the list of makes with at least 50 occurrences
valid_makes = counts[counts >= 50].index
```

```python
# Filter DataFrame to keep only valid makes
df = df[df['Make'].isin(valid_makes)]
```

```python
# Verify top makes and remaining dataset shape
print(df['Make'].value_counts().head(15))
print(f"Remaining rows after make filtering: {len(df)}")
```

### Cleaning other columns
- there are other columns containing data that might be related to the outcome of an accident. We list a few here:
- Engine.Type
- Weather.Condition
- Number.of.Engines
- Purpose.of.flight
- Broad.phase.of.flight

Inspect and identify potential cleaning tasks in each of the above columns. Execute those cleaning tasks. 

**Note**: You do not necessarily need to impute or drop NaNs here.

```python
# Load state codes
states = pd.read_csv('USState_Codes.csv')
```

```python
# Extract postal abbreviation
df['State_Code'] = df['Location'].str.split(',').str[-1].str.strip()
```

```python
# Merge on abbreviation to obtain full state names
df.merge(states, left_on='State_Code', right_on='Abbreviation', how='left')
```

```python
# Weather.Condition: Standardize casing and group placeholders
df['Weather.Condition'] = df['Weather.Condition'].astype(str).str.strip().str.upper()
df['Weather.Condition'] = df['Weather.Condition'].replace({
    'UNK': 'UNKNOWN',
    'NAN': np.nan
})

# Engine.Type: Standardize casing and collapse unknown variants
df['Engine.Type'] = df['Engine.Type'].astype(str).str.strip().str.title()
df['Engine.Type'] = df['Engine.Type'].replace({
    'Unk': 'Unknown',
    'None': 'Unknown',
    'Lr': 'Unknown',
    'Nan': np.nan
})

# Number.of.Engines-- Replace invalid engine count (0 engines) with NaN
# Airplanes must have at least 1 engine; 0 is an error
df.loc[df['Number.of.Engines'] == 0, 'Number.of.Engines'] = np.nan

# Broad.phase.of.flight: Standardize text formatting
df['Broad.phase.of.flight'] = df['Broad.phase.of.flight'].astype(str).str.strip().str.title()
df['Broad.phase.of.flight'] = df['Broad.phase.of.flight'].replace({'Nan': np.nan})

# Purpose.of.flight: Standardize text formatting
df['Purpose.of.flight'] = df['Purpose.of.flight'].astype(str).str.strip().str.title()
df['Purpose.of.flight'] = df['Purpose.of.flight'].replace({'Nan': np.nan})
```

```python
for col in ['Weather.Condition', 'Engine.Type', 'Number.of.Engines', 'Broad.phase.of.flight', 'Purpose.of.flight']:
    print(f"=== {col} ===")
    print(df[col].value_counts(dropna=False).head(6))
    print()
```
